## Rename response variable


```python
for df_name in ('llama_df', 'qwen_df', 'mistral_df'):
    if df_name in globals():
        df = globals()[df_name]
        df.rename(columns={'response': 'response_original', 'response_clean': 'response'}, inplace=True)
        print(f"{df_name} columns:", df.columns.tolist())
```

```python
# Count NAs in the 'response' column for available DataFrames
df_names = ['llama_df', 'qwen_df', 'mistral_df']

for name in df_names:
    if name in globals():
        d = globals()[name]
        if 'response' in d.columns:
            na_count = d['response'].isna().sum()
            total = len(d)
            pct = na_count / total * 100 if total > 0 else 0
            print(f"{name}: NA count = {na_count:,} / {total:,} ({pct:.2f}%)")
        else:
            print(f"{name}: no 'response' column")
    else:
        print(f"{name}: not defined")
```
